Data Quality Checks

```python
## Print the number of null values in each column
print(df.isnull().sum())
```

```python
## Check for duplicate records in the dataset
duplicate_records = df.duplicated().sum()
print(f"Total Duplicate Records: {duplicate_records}")
```

```python
## Check if the same PatientID was generated for multiple patients
id_duplicates = df.duplicated(subset=['PatientID']).sum()
print(f"Duplicated PatientIDs Found: {id_duplicates}")
```

```python
## Clean any trailing or hidden spaces
string_columns = df.select_dtypes(include=['object', 'string']).columns
for col in string_columns:
    df[col] = df[col].astype(str).str.strip()
```

Calculate Overall Baseline Churn Rate

```python
## Total churn rate
total_patients = len(df)
churned_patients = df['Churned'].sum()
overall_churn_rate = (churned_patients / total_patients) * 100

print(f"Total Patients: {total_patients}")
print(f"Total Churned: {churned_patients}")
print(f"Overall Churn Rate: {overall_churn_rate:.2f}%")
```

Breakdown Churn by Specialty


```python
## Group by specialty
specialty_summary = df.groupby('Specialty')['Churned'].agg(
    total_patients='count',
    churned_patients='sum',
    churn_rate='mean'
).reset_index()

## Covert churn_rate to percentage
specialty_summary['churn_rate'] = (specialty_summary['churn_rate'] * 100).round(2)

## Sort descending to see highest churn departments first
specialty_summary = specialty_summary.sort_values(by='churn_rate', ascending=False)
print(specialty_summary)
```

Breakdown Churn by Insurance Type


```python
## Group by insurance
insurance_summary = df.groupby('Insurance_Type')['Churned'].agg(
    total_patients='count',
    churned_patients='sum',
    churn_rate='mean'
).reset_index()

## Convert churn_rate to percentage
insurance_summary['churn_rate'] = (insurance_summary['churn_rate'] * 100).round(2)

## Sort descending
insurance_summary = insurance_summary.sort_values(by='churn_rate', ascending=False)
print(insurance_summary)
```

Wait-Time Satisfaction & Missed Appointments

```python
## Compare average wait-time satisfaction by churn status
wait_time_impact = df.groupby('Churned')['Wait_Time_Satisfaction'].agg(['mean', 'median', 'std']).reset_index()
print(f"---Wait Time Satisfaction by Churn Status---")
print(wait_time_impact)

## Churn rate by count of missed appointments
missed_appointments_impact = df.groupby('Missed_Appointments')['Churned'].agg(
    total_patients='count',
    churned_patients='sum',
    churn_rate='mean'
).reset_index()
missed_appointments_impact['churn_rate'] = (missed_appointments_impact['churn_rate'] * 100).round(2)

print(f"---Churn Rate by Missed Appointments---")
print(missed_appointments_impact)
```

Administrative Friction (Billing & Patient Portal)

```python
## Churn rate by billing issues
billing_impact = df.groupby('Billing_Issues')['Churned'].agg(
    total_patients='count',
    churned_patients='sum',
    churn_rate='mean'
).reset_index()
billing_impact['churn_rate'] = (billing_impact['churn_rate'] * 100).round(2)

print(f"---Churn Rate by Billing Issue---")
print(billing_impact)

## Churn rate by patient portal usage
portal_impact = df.groupby('Portal_Usage')['Churned'].agg(
    total_patients='count',
    churned_patients='sum',
    churn_rate='mean'
).reset_index()
portal_impact['churn_rate'] = (portal_impact['churn_rate'] * 100).round(2)

print(f"---Churn Rate by Patient Portal Usage---")
print(portal_impact)
```

Distance & Geographical Access

```python
## Segment facility distance 
bins = [0, 10, 25, 50]
labels = ['0-10 miles', '10-25 miles', '25+ miles']
df['distance_band'] = pd.cut(df['Distance_To_Facility_Miles'], bins=bins, labels=labels)

distance_impact = df.groupby('distance_band', observed=False)['Churned'].agg(
    total_patients='count',
    churned_patients='sum',
    churn_rate='mean'
).reset_index()
distance_impact['churn_rate']= (distance_impact['churn_rate'] * 100).round(2)

print(f"---Churn Rate by Distance---")
print(distance_impact)
```

High-Risk Patient Profiling


```python
## Define high-risk criteria
high_risk_flag = (df['Portal_Usage'] == 0) & (
    (df['Billing_Issues'] == 1) | (df['Missed_Appointments'] >= 3)
)

## Split dataset into high-risk vs standard
high_risk_df = df[high_risk_flag]
standard_df = df[~high_risk_flag]

## Compare churn rates
print(
    f"High-Risk Segment Size: {len(high_risk_df)} patients"
    f" ({len(high_risk_df)/len(df)*100:.1f}% of total)"
)
print(f"High-Risk Churn Rate: {high_risk_df['Churned'].mean()*100:.2f}%")
print(f"Standard Segment Churn Rate: {standard_df['Churned'].mean()*100:.2f}%")
```

Quantify Financial & Visit Impact

```python
## Total revenue exposure across churned high-risk patients
churned_high_risk = high_risk_df[high_risk_df['Churned'] == 1]

total_lost_oop_revenue = churned_high_risk['Avg_Out_Of_Pocket_Cost'].sum()
total_lost_visits = churned_high_risk['Visits_Last_Year'].sum()

print(f"---High-Risk Churned Patient Impact---")
print(f"Total Churned High-Risk Patients: {len(churned_high_risk)}")
print(f"Total Annual Lost OOP Revenue: ${total_lost_oop_revenue:,.2f}")
print(f"Total Annual Lost Visit Volume: {total_lost_visits:,} visits")
```

Create Clean Excel Export

```python

## Create Executive Summary Table
df_summary = pd.DataFrame({
    'Metric': [
        'Total Patients',
        'Total Churned Patients',
        'Overall Churn Rate',
        'High-Risk Segment Size',
        'High-Risk Churn Rate',
        'Total Lost OOP Revenue (High-Risk Churned)',
        'Total Lost Annual Visits (High-Risk Churned)',
    ],
    'Value': [
        len(df),
        df['Churned'].sum(),
        f'{(df["Churned"].mean() * 100):.2f}%',
        len(df[high_risk_flag]),
        f'{(df[high_risk_flag]["Churned"].mean() * 100):.2f}%',
        f'${df[high_risk_flag & (df["Churned"] == 1)]["Avg_Out_Of_Pocket_Cost"].sum():,.2f}',
        f'{df[high_risk_flag & (df["Churned"] == 1)]["Visits_Last_Year"].sum():,}',
    ],
})

## Create Specialty Summary Table
df_specialty = (
    df.groupby('Specialty')['Churned']
    .agg(total_patients='count', churned_patients='sum', churn_rate='mean')
    .reset_index()
)
df_specialty['churn_rate_%'] = (df_specialty['churn_rate'] * 100).round(2)
df_specialty = df_specialty.drop(columns=['churn_rate']).sort_values(
    by='churn_rate_%', ascending=False
)

## Create Insurance Summary Table
df_insurance = (
    df.groupby('Insurance_Type')['Churned']
    .agg(total_patients='count', churned_patients='sum', churn_rate='mean')
    .reset_index()
)
df_insurance['churn_rate_%'] = (df_insurance['churn_rate'] * 100).round(2)
df_insurance = df_insurance.drop(columns=['churn_rate']).sort_values(
    by='churn_rate_%', ascending=False
)

## Export to Excel File
excel_filename = r"C:\Users\Edumo\Desktop\patient_churn_cleaned_export.xlsx"

with pd.ExcelWriter(excel_filename, engine='openpyxl') as writer:
    df_summary.to_excel(writer, sheet_name='Executive Summary', index=False)
    df_specialty.to_excel(writer, sheet_name='Specialty Metrics', index=False)
    df_insurance.to_excel(writer, sheet_name='Insurance Metrics', index=False)
    df[high_risk_flag].to_excel(
        writer, sheet_name='High-Risk Patient Group', index=False)
    df.to_excel(writer, sheet_name='Cleaned Patient Data', index=False)

print(f'Excel workbook successfully exported to {excel_filename}')
```
